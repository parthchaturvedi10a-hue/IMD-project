```python
import os
import inspect
import zipfile
import numpy as np
import pandas as pd
import torch
import transformers

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
)
from torch.utils.data import Dataset
from transformers import (
    DistilBertTokenizer,
    DistilBertForSequenceClassification,
    Trainer,
    TrainingArguments,
)

print("Transformers version:", transformers.__version__)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("WARNING: no GPU. Set Runtime -> Change runtime type -> T4 GPU")
#%% md
## 3. Upload and extract the dataset zip
```

```python
from google.colab import files

uploaded = files.upload()
zip_file = list(uploaded.keys())[0]
print("Uploaded file:", zip_file)

extract_path = "email_dataset"
os.makedirs(extract_path, exist_ok=True)
with zipfile.ZipFile(zip_file, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("\nExtracted files:")
for root, dirs, file_list in os.walk(extract_path):
    for file in file_list:
        print(os.path.join(root, file))
```

```python
csv_files = []
for root, dirs, file_list in os.walk(extract_path):
    for file in file_list:
        if file.lower().endswith(".csv"):
            csv_files.append(os.path.join(root, file))

# FIX 1: remove the duplicate combined file
csv_files = [f for f in csv_files if "phishing_email" not in f]

print("CSV files used:")
for file in csv_files:
    print(" ", file)
assert len(csv_files) == 6, "Expected 6 source files (CEAS, Enron, Ling, Nazario, Nigerian, SpamAssassin)"
```

```python
all_data = []

for file in csv_files:
    print("\nReading:", file)
    try:
        df = pd.read_csv(file, encoding="utf-8", on_bad_lines="skip")
    except Exception:
        df = pd.read_csv(file, encoding="latin-1", on_bad_lines="skip")

    print("  Rows:", len(df), "| Columns:", list(df.columns))

    if "subject" in df.columns and "body" in df.columns:
        df["subject"] = df["subject"].fillna("").astype(str)
        df["body"] = df["body"].fillna("").astype(str)
        df["text"] = "Subject: " + df["subject"] + "\nBody: " + df["body"]
    elif "body" in df.columns:
        df["text"] = "Subject: \nBody: " + df["body"].fillna("").astype(str)
    elif "text" in df.columns:
        df["text"] = df["text"].fillna("").astype(str)
    else:
        print("  No usable text column, skipping")
        continue

    if "label" not in df.columns:
        print("  No label column, skipping")
        continue

    temp = df[["text", "label"]].copy()
    temp["source"] = os.path.basename(file).replace(".csv", "")
    all_data.append(temp)

data = pd.concat(all_data, ignore_index=True)
print("\nCOMBINED rows:", len(data))
print(data["label"].value_counts())
```

```python
data = data.dropna(subset=["text", "label"])
data["text"] = data["text"].astype(str)
data = data[data["text"].str.strip() != ""]

data["label"] = pd.to_numeric(data["label"], errors="coerce")
data = data.dropna(subset=["label"])
data["label"] = data["label"].astype(int)
data = data[data["label"].isin([0, 1])]

data = data.drop_duplicates(subset=["text"]).reset_index(drop=True)

print("AFTER CLEANING rows:", len(data), " (should be about 82,000)")
print(data["label"].value_counts())
print("\nEmails per source and label:")
print(data.groupby(["source", "label"]).size().unstack(fill_value=0))

# stratify by source + label so every dataset is represented in each split
strat = data["source"] + "_" + data["label"].astype(str)
train_data, temp_data = train_test_split(data, test_size=0.20, random_state=42, stratify=strat)
val_data, test_data = train_test_split(
    temp_data, test_size=0.50, random_state=42,
    stratify=temp_data["source"] + "_" + temp_data["label"].astype(str))

train_data = train_data.reset_index(drop=True)
val_data = val_data.reset_index(drop=True)
test_data = test_data.reset_index(drop=True)

print("\nTraining:", len(train_data), "| Validation:", len(val_data), "| Testing:", len(test_data))
```

```python
MAX_LENGTH = 256
tokenizer = DistilBertTokenizer.from_pretrained("distilbert-base-uncased")

class EmailDataset(Dataset):
    def __init__(self, dataframe, tokenizer, max_length=MAX_LENGTH):
        self.texts = dataframe["text"].tolist()
        self.labels = dataframe["label"].tolist()
        self.tokenizer = tokenizer
        self.max_length = max_length

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, index):
        encoding = self.tokenizer(
            self.texts[index],
            truncation=True,
            max_length=self.max_length,
            padding="max_length",
            return_tensors="pt",
        )
        return {
            "input_ids": encoding["input_ids"].flatten(),
            "attention_mask": encoding["attention_mask"].flatten(),
            "labels": torch.tensor(int(self.labels[index]), dtype=torch.long),
        }

train_dataset = EmailDataset(train_data, tokenizer)
val_dataset = EmailDataset(val_data, tokenizer)
test_dataset = EmailDataset(test_data, tokenizer)
print("Datasets created.")
```

```python
model = DistilBertForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=2)
model.to(device)

def compute_metrics(pred):
    labels = pred.label_ids
    predictions = np.argmax(pred.predictions, axis=1)
    precision, recall, f1, _ = precision_recall_fscore_support(
        labels, predictions, average="binary", zero_division=0)
    return {"accuracy": accuracy_score(labels, predictions),
            "precision": precision, "recall": recall, "f1": f1}

training_parameters = {
    "output_dir": "./distilbert_phishing",
    "num_train_epochs": 2,              # FIX 2: was 3; val loss rose at epoch 3
    "learning_rate": 2e-5,
    "per_device_train_batch_size": 16,
    "per_device_eval_batch_size": 32,
    "weight_decay": 0.01,
    "warmup_ratio": 0.06,
    "save_strategy": "epoch",
    "load_best_model_at_end": True,
    "metric_for_best_model": "f1",
    "greater_is_better": True,
    "logging_steps": 100,
    "save_total_limit": 2,
    "report_to": "none",
    "fp16": torch.cuda.is_available(),
}

available = inspect.signature(TrainingArguments.__init__).parameters
if "eval_strategy" in available:
    training_parameters["eval_strategy"] = "epoch"
elif "evaluation_strategy" in available:
    training_parameters["evaluation_strategy"] = "epoch"

# keep only the settings this Transformers version supports
training_parameters = {k: v for k, v in training_parameters.items() if k in available}
training_args = TrainingArguments(**training_parameters)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    compute_metrics=compute_metrics,
)
print("Trainer ready.")
```

```python
trainer.train()
```

```python
test_results = trainer.evaluate(test_dataset)
print("TEST RESULTS")
for key, value in test_results.items():
    print(f"  {key}: {value:.4f}" if isinstance(value, float) else f"  {key}: {value}")

predictions = trainer.predict(test_dataset)
predicted_labels = np.argmax(predictions.predictions, axis=1)
actual_labels = test_data["label"].values

print("\nCLASSIFICATION REPORT")
print(classification_report(actual_labels, predicted_labels,
                            target_names=["Legitimate", "Phishing"], digits=4, zero_division=0))

print("CONFUSION MATRIX")
print(confusion_matrix(actual_labels, predicted_labels))
print("[[True Legitimate, False Phishing]")
print(" [False Legitimate, True Phishing]]")

# accuracy per source dataset: shows where the model is weakest
test_data["pred"] = predicted_labels
print("\nACCURACY PER SOURCE")
print(test_data.groupby("source").apply(lambda g: (g["pred"] == g["label"]).mean()).round(4))
```

```python
def predict_email(body, subject=""):
    # FIX 3: same input format as training
    text = f"Subject: {subject}\nBody: {body}"
    model.eval()
    inputs = tokenizer(text, return_tensors="pt", truncation=True,
                       max_length=MAX_LENGTH, padding=True)
    inputs = {k: v.to(model.device) for k, v in inputs.items()}
    with torch.no_grad():
        probabilities = torch.softmax(model(**inputs).logits, dim=1)[0]
    prediction = int(torch.argmax(probabilities))
    label = "PHISHING" if prediction == 1 else "LEGITIMATE"
    return label, float(probabilities[prediction])

tests = [
    ("Urgent: account suspended",
     "Your bank account will be suspended. Click the link below immediately to verify your account. "
     "Please provide your password, credit card number and personal information."),
    ("Meeting tomorrow",
     "Hi John, the meeting has been scheduled for tomorrow at 10 AM. Please bring the project report. Regards, David"),
]
for subject, body in tests:
    label, confidence = predict_email(body, subject)
    print(f"{subject!r:35} -> {label} ({confidence * 100:.2f}%)")
```

```python
%%writefile phishing_analyzer.py
"""
Phishing Risk Analyzer: DistilBERT + Explainable AI + URL intelligence.

    from phishing_analyzer import PhishingAnalyzer, make_distilbert_scorer
    scorer = make_distilbert_scorer("./distilbert_phishing_model")
    analyzer = PhishingAnalyzer(scorer)
    report = analyzer.analyze(subject="...", body="...", sender="...")
    print(report.text())            # boxed console report
    display(HTML(report.html()))    # rich report in Colab

A "scorer" is any function list[str] -> np.ndarray of phishing LOG-ODDS
(logit_phishing - logit_legit). Log-odds stay informative when the model
is saturated at ~100% confidence; probabilities don't.

Components
  1. Text model            DistilBERT log-odds -> probability
  2. XAI                   word occlusion: remove each word, measure score change
  3. Counterfactual        smallest set of top words whose removal flips the verdict
  4. URL analyzer          13 technical URL checks
  5. Sender analyzer       brand spoofing, lookalike and freemail checks
  6. Indicators            urgency / credential / threat / financial language
  7. Attack type           rule-based taxonomy (dataset has no attack-type labels)
  8. Risk engine           combined score, risk level, auto vs human-review routing
  9. Robustness            prediction stability under small text perturbations
"""
from __future__ import annotations
import html as _html, math, re
from dataclasses import dataclass, field
import numpy as np

# =====================================================================
# 1. MODEL ADAPTER
# =====================================================================
def make_distilbert_scorer(model_dir="./distilbert_phishing_model", max_length=256,
                           batch_size=32, device=None):
    """Wrap a fine-tuned (Distil)BERT classifier (label 1 = phishing)."""
    import torch
    from transformers import AutoTokenizer, AutoModelForSequenceClassification
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    tok = AutoTokenizer.from_pretrained(model_dir)
    model = AutoModelForSequenceClassification.from_pretrained(model_dir).to(device).eval()

    @torch.no_grad()
    def score(texts):
        out = []
        for i in range(0, len(texts), batch_size):
            enc = tok(texts[i:i + batch_size], truncation=True, max_length=max_length,
                      padding=True, return_tensors="pt").to(device)
            logits = model(**enc).logits.float()
            out.append((logits[:, 1] - logits[:, 0]).cpu().numpy())
        return np.concatenate(out)
    return score

def format_email(subject, body):
    """Must match the training format in distillbert.ipynb."""
    return f"Subject: {subject}\nBody: {body}"

def sigmoid(x): return 1 / (1 + np.exp(-np.clip(x, -50, 50)))

# =====================================================================
# 2. LEXICONS
# =====================================================================
def _lex(words): return re.compile(r"\b(?:" + "|".join(words) + r")\b", re.I)

INDICATORS = {
    "Urgency language": _lex([
        "urgent", "urgently", "immediately", "asap", "within \\d+ (?:hours|days)",
        "act now", "right away", "expires?", "expired", "final notice", "last chance",
        "limited time", "deadline", "today only", "as soon as possible"]),
    "Credential request": _lex([
        "password", "passcode", "verify", "verification", "confirm your", "login",
        "log in", "sign in", "username", "credentials", "pin", "ssn",
        "social security", "account details", "update your (?:account|information|details)"]),
    "Threat / account suspension": _lex([
        "suspended", "suspend", "locked", "unauthorized", "unusual activity",
        "compromised", "disabled", "terminated", "security alert", "restricted",
        "deactivated", "blocked", "legal action"]),
    "Financial lure": _lex([
        "million", "inheritance", "lottery", "prize", "winner", "wire transfer",
        "beneficiary", "compensation", "claim your", "refund", "cash",
        "\\$\\s?\\d[\\d,]*", "usd", "bank account"]),
    "Generic greeting": _lex([
        "dear customer", "dear user", "dear member", "dear friend", "dear sir",
        "dear account holder", "valued customer", "dear beneficiary", "dear client"]),
    "Call to action link": _lex(["click here", "click below", "click the link",
                                 "follow the link", "tap here", "open the attachment"]),
}

ATTACK_TYPES = {
    "Credential Harvesting": (["password", "login", "log in", "sign in", "verify your",
        "confirm your", "credentials", "mailbox", "webmail", "username",
        "validate", "re-activate", "reactivate"],
        "Fake login page designed to steal usernames and passwords."),
    "Account Takeover / Security Alert": (["suspended", "locked", "unusual activity",
        "unauthorized", "security alert", "reset your password", "compromised",
        "deactivated", "sign-in attempt"],
        "Fake security warning that pressures you to 'secure' your account."),
    "Advance-Fee / Financial Scam": (["million", "inheritance", "beneficiary", "next of kin",
        "barrister", "transfer of", "fund", "lottery", "prize", "winner",
        "consignment", "diplomat", "compensation"],
        "Promises large sums of money in exchange for fees or personal details."),
    "Payment / Invoice Fraud": (["invoice", "payment", "wire", "bank details", "overdue",
        "receipt", "transaction", "refund", "billing", "remittance"],
        "Fake bill, receipt or refund aimed at payments or card details."),
    "Delivery Scam": (["package", "parcel", "shipment", "delivery", "tracking",
        "courier", "dhl", "fedex", "ups", "customs"],
        "Fake shipping notice that leads to a malicious link or fee."),
    "Spam / Marketing": (["unsubscribe", "discount", "buy now", "special offer", "deal",
        "viagra", "pills", "pharmacy", "casino", "save up to", "% off", "free shipping"],
        "Unsolicited bulk advertising."),
}

BRANDS = ["paypal", "amazon", "apple", "microsoft", "netflix", "chase", "wellsfargo",
          "bankofamerica", "dhl", "fedex", "ups", "irs", "ebay", "google", "facebook",
          "instagram", "linkedin", "outlook", "office365", "visa", "mastercard", "hsbc",
          "citibank", "dropbox", "docusign", "adobe", "whatsapp", "sbi", "hdfc", "icici"]
SHORTENERS = {"bit.ly", "tinyurl.com", "goo.gl", "t.co", "ow.ly", "is.gd", "buff.ly",
              "rebrand.ly", "cutt.ly", "shorturl.at", "tiny.cc", "rb.gy", "t.ly"}
SUSPICIOUS_TLDS = {"ru", "cn", "tk", "top", "xyz", "info", "biz", "ml", "ga", "cf", "gq",
                   "work", "click", "link", "cc", "ws", "su", "zip", "mov", "rest", "cam"}
FREEMAIL = {"gmail.com", "yahoo.com", "hotmail.com", "outlook.com", "aol.com", "mail.com",
            "yandex.ru", "live.com", "msn.com", "gmx.com", "rediffmail.com", "protonmail.com",
            "yahoo.co.in", "yahoo.co.uk", "icloud.com"}
URL_KEYWORDS = ["login", "signin", "verify", "secure", "account", "update", "confirm",
                "banking", "webscr", "password", "wallet"]
LEET = str.maketrans({"0": "o", "1": "l", "3": "e", "4": "a", "5": "s", "7": "t", "@": "a"})

URL_RE = re.compile(r"(?:https?://|www\.)[^\s<>\"')\]]+", re.I)
SENDER_RE = re.compile(r"([\w.+-]+)@([\w-]+(?:\.[\w-]+)+)")

# =====================================================================
# 3. URL + SENDER ANALYZERS
# =====================================================================
def _host(url):
    u = re.sub(r"^https?://", "", url, flags=re.I).split("/")[0].split("?")[0]
    return u.split("@")[-1].split(":")[0].lower()

def _registered(host):
    p = host.split(".")
    if len(p) >= 3 and p[-2] in {"co", "com", "org", "gov", "ac", "net"} and len(p[-1]) == 2:
        return ".".join(p[-3:])
    return ".".join(p[-2:])

def _brand_of(registered):
    return registered.split(".")[0].replace("-", "")

def _lookalike(name):
    """Brand hidden by character swaps: paypa1-secure -> paypal, rnicrosoft -> microsoft."""
    d = name.translate(LEET).replace("rn", "m")
    for b in BRANDS:
        if len(b) >= 4 and b in d and b not in name:
            return b
    return None

def analyze_url(url):
    """Return (list of (finding, severity 0-1)) for one URL."""
    f, host = [], _host(url)
    reg = _registered(host)
    if re.fullmatch(r"\d{1,3}(?:\.\d{1,3}){3}", host):
        f.append(("Uses an IP address instead of a domain name", 0.8))
    if "@" in re.sub(r"^https?://", "", url).split("/")[0]:
        f.append(("Contains '@', which hides the real destination", 0.8))
    if "xn--" in host:
        f.append(("Punycode domain (possible look-alike characters)", 0.7))
    if reg in SHORTENERS:
        f.append(("URL shortener hides the real destination", 0.5))
    if url.lower().startswith("http://"):
        f.append(("Not encrypted (http instead of https)", 0.25))
    if len(url) > 100:
        f.append((f"Unusually long URL ({len(url)} characters)", 0.25))
    if host.count(".") >= 4:
        f.append((f"Excessive subdomains ({host.count('.')} dots)", 0.4))
    if host.rsplit(".", 1)[-1] in SUSPICIOUS_TLDS:
        f.append((f"High-risk top-level domain (.{host.rsplit('.', 1)[-1]})", 0.4))
    if reg.split(".")[0].count("-") >= 2:
        f.append(("Many hyphens in domain name", 0.3))
    if re.search(r":\d{2,5}", re.sub(r"^https?://", "", url).split("/")[0]):
        f.append(("Non-standard port number", 0.4))
    # brand used outside its real domain, e.g. paypal.com.secure-login.ru
    reg_brand = _brand_of(reg)
    for b in BRANDS:
        if b in url.lower().replace("-", "") and b != reg_brand:
            f.append((f"Mentions '{b}' but the real domain is {reg}", 0.75)); break
    # look-alike, e.g. paypa1.com, rnicrosoft.com
    la = _lookalike(reg_brand)
    if la:
        f.append((f"Look-alike of '{la}' ({reg})", 0.9))
    kws = [k for k in URL_KEYWORDS if k in url.lower()]
    if kws:
        f.append((f"Sensitive keywords in URL: {', '.join(kws[:3])}", 0.2))
    return f

def analyze_sender(sender):
    if not sender or not sender.strip():
        return [], ""
    f, m = [], SENDER_RE.search(sender)
    if not m:
        return [("Sender address could not be parsed", 0.2)], ""
    dom = m.group(2).lower(); reg = _registered(dom)
    display = sender.split("<")[0].lower().replace(" ", "") if "<" in sender else ""
    for b in BRANDS:
        if b in display and b != _brand_of(reg):
            f.append((f"Display name claims '{b}' but email comes from {dom}", 0.8)); break
    la = _lookalike(_brand_of(reg))
    if la:
        f.append((f"Look-alike sender domain imitating '{la}' ({dom})", 0.9))
    if reg in FREEMAIL and any(b in display for b in BRANDS):
        f.append(("Brand email sent from a free webmail account", 0.6))
    if sum(c.isdigit() for c in dom) >= 3:
        f.append(("Sender domain contains many digits", 0.3))
    if dom.rsplit(".", 1)[-1] in SUSPICIOUS_TLDS:
        f.append((f"Sender uses high-risk domain (.{dom.rsplit('.', 1)[-1]})", 0.4))
    return f, dom

def combine(severities):
    """Noisy-OR: independent findings accumulate risk."""
    return 1 - math.prod(1 - s for s in severities) if severities else 0.0

# =====================================================================
# 4. REPORT OBJECT
# =====================================================================
@dataclass
class Report:
    sender: str; subject: str; body: str
    prob: float; logodds: float
    verdict: str; risk_score: float; risk_level: str; decision: str
    word_attr: dict = field(default_factory=dict)       # word -> log-odds contribution
    top_phish: list = field(default_factory=list)
    top_legit: list = field(default_factory=list)
    indicators: dict = field(default_factory=dict)      # name -> matched phrases
    urls: list = field(default_factory=list)            # (url, findings)
    url_risk: float = 0.0
    sender_findings: list = field(default_factory=list)
    sender_risk: float = 0.0
    attack_type: str = ""; attack_conf: float = 0.0; attack_desc: str = ""
    counterfactual: dict = field(default_factory=dict)
    robustness: list = field(default_factory=list)
    reasons: list = field(default_factory=list)
    recommendation: str = ""

    # ---------------- console report ----------------
    def text(self, width=56):
        W = width
        def line(s="", indent="   "):
            rows, s = [], str(s)
            while True:
                wide = sum(1 for c in s[:W - 4] if ord(c) >= 0x1F000)
                n = W - 4 - wide
                if len(s) <= n:
                    rows.append(f"║ {s}{' ' * (n - len(s))} ║"); break
                cut = s.rfind(" ", 0, n)
                cut = cut if cut > n // 2 else n
                rows.append(f"║ {s[:cut]}{' ' * (n - cut)} ║"); s = indent + s[cut:].lstrip()
            return "\n".join(rows)
        sep, top, bot = "╠" + "═" * (W - 2) + "╣", "╔" + "═" * (W - 2) + "╗", "╚" + "═" * (W - 2) + "╝"
        icon = {"PHISHING": "⚠ PHISHING", "LEGITIMATE": "✔ LEGITIMATE"}[self.verdict]
        out = [top, line("PHISHING RISK ANALYSIS".center(W - 4)), sep,
               line(f"Prediction:   {icon}"),
               line(f"Confidence:   {max(self.prob, 1 - self.prob) * 100:.1f}%"),
               line(f"Risk level:   {self.risk_level} (score {self.risk_score:.2f})"),
               line(f"Decision:     {self.decision}")]
        if self.verdict == "PHISHING":
            out.append(line(f"Attack type:  {self.attack_type} ({self.attack_conf:.0%})"))
        hdr = "WHY WAS IT FLAGGED?" if self.verdict == "PHISHING" else "WHY WAS IT CONSIDERED SAFE?"
        out += [sep, line(hdr), line("(change in log-odds when the word is removed)"), line()]
        mx = max([abs(v) for _, v in self.top_phish + self.top_legit] or [1]) or 1
        red = [("🔴", w, v) for w, v in self.top_phish]
        green = [("🟢", w, v) for w, v in self.top_legit]
        rows = red[:6] + green[:2] if self.verdict == "PHISHING" else green[:5] + red[:3]
        for ic, w, v in rows:
            out.append(line(f"{ic} {w[:16]:<16} {v:+6.2f} {'█' * max(1, int(12 * abs(v) / mx))}"))
        if self.indicators:
            out += [sep, line("SUSPICIOUS INDICATORS"), line()]
            for k, v in self.indicators.items():
                out.append(line(f"✓ {k}: {', '.join(v[:2])}"))
        if self.urls or self.sender_findings:
            out += [sep, line(f"URL & SENDER ANALYSIS  (URL risk {self.url_risk:.2f})"), line()]
            for u, fs in self.urls[:3]:
                out.append(line(f"🔗 {u[:W-8]}"))
                for d, _ in fs[:3]:
                    out.append(line(f"   - {d}"))
            for d, _ in self.sender_findings:
                out.append(line(f"✉ {d}"))
        cf = self.counterfactual
        if cf.get("removed"):
            out += [sep, line("WHAT-IF (counterfactual)"), line()]
            out.append(line(f"Removing: {', '.join(cf['removed'][:6])}"))
            out.append(line(f"Phishing prob {cf['prob_before']:.1%} -> {cf['prob_after']:.1%}"
                            + ("  (verdict flips)" if cf["flipped"] else "")))
        out += [sep, line("RECOMMENDATION")]
        for chunk in re.findall(r".{1,%d}(?:\s|$)" % (W - 5), self.recommendation):
            out.append(line(chunk.strip()))
        out.append(bot)
        return "\n".join(out)

    # ---------------- HTML report ----------------
    def html(self):
        e = _html.escape
        color = {"HIGH": "#d93025", "MEDIUM": "#e37400", "LOW": "#188038"}[self.risk_level]
        mx = max([abs(v) for v in self.word_attr.values()] or [1]) or 1

        def highlight(text):
            parts = []
            for tok in re.findall(r"\S+|\s+", text):
                k = _key(tok); v = self.word_attr.get(k, 0)
                a = min(abs(v) / mx, 1) * 0.75
                if k and a > 0.04:
                    bg = f"rgba(217,48,37,{a:.2f})" if v > 0 else f"rgba(24,128,56,{a:.2f})"
                    parts.append(f'<span title="{v:+.2f}" style="background:{bg};border-radius:3px;padding:0 1px">{e(tok)}</span>')
                else:
                    parts.append(e(tok))
            return "".join(parts).replace("\n", "<br>")

        def bars(items, col):
            return "".join(
                f'<div style="display:flex;align-items:center;gap:8px;margin:3px 0">'
                f'<code style="width:150px;overflow:hidden;white-space:nowrap;text-overflow:ellipsis" title="{e(w)}">{e(w[:24])}</code>'
                f'<div style="background:{col};height:12px;width:{int(160 * abs(v) / mx)}px;border-radius:2px"></div>'
                f'<span style="font-size:12px">{v:+.2f}</span></div>' for w, v in items)

        ind = "".join(f"<li><b>{e(k)}</b>: {e(', '.join(v[:3]))}</li>" for k, v in self.indicators.items()) or "<li>None detected</li>"
        urls = "".join(
            f"<li><code>{e(u[:90])}</code><ul>" + "".join(f"<li>{e(d)} <small>(severity {s:.2f})</small></li>" for d, s in fs)
            + ("<li><small>no technical red flags</small></li>" if not fs else "") + "</ul></li>"
            for u, fs in self.urls[:6]) or "<li>No links found</li>"
        snd = "".join(f"<li>{e(d)}</li>" for d, _ in self.sender_findings) or "<li>No sender red flags (or no sender given)</li>"
        reasons = "".join(f"<li>{e(r)}</li>" for r in self.reasons)
        cf = self.counterfactual
        cf_html = (f"<p>Removing <b>{len(cf['removed'])}</b> word(s) "
                   f"(<code>{e(', '.join(cf['removed']))}</code>) changes the phishing probability from "
                   f"<b>{cf['prob_before']:.1%}</b> to <b>{cf['prob_after']:.1%}</b>"
                   f"{' and <b>flips the verdict</b>' if cf['flipped'] else ' (verdict does not flip: the decision relies on many signals)'}.</p>"
                   if cf.get("removed") else "<p>Not applicable.</p>")
        rob = "".join(f"<tr><td>{e(n)}</td><td>{p:.1%}</td><td>{'✔' if (p >= .5) == (self.prob >= .5) else '✘ flipped'}</td></tr>"
                      for n, p in self.robustness)
        card = "border:1px solid #8884;border-radius:10px;padding:12px 16px;margin:10px 0"
        return f"""
<div style="font-family:system-ui,sans-serif;max-width:900px;line-height:1.45">
 <div style="{card};border-left:6px solid {color}">
  <div style="font-size:13px;opacity:.7">PHISHING RISK ANALYSIS</div>
  <div style="font-size:26px;font-weight:700;color:{color}">{'⚠ PHISHING' if self.verdict == 'PHISHING' else '✔ LEGITIMATE'}</div>
  <div style="display:flex;gap:28px;flex-wrap:wrap;margin-top:6px">
   <div>Confidence<br><b>{max(self.prob, 1 - self.prob) * 100:.1f}%</b></div>
   <div>Risk level<br><b style="color:{color}">{self.risk_level}</b> ({self.risk_score:.2f})</div>
   <div>Decision<br><b>{e(self.decision)}</b></div>
   {'<div>Attack type<br><b>' + e(self.attack_type) + f'</b> ({self.attack_conf:.0%})</div>' if self.verdict == 'PHISHING' else ''}
  </div>
  {'<div style="font-size:13px;margin-top:6px;opacity:.8">' + e(self.attack_desc) + '</div>' if self.verdict == 'PHISHING' else ''}
 </div>
 <div style="{card}"><b>Main reasons</b><ol>{reasons}</ol></div>
 <div style="{card}"><b>{'Why was it flagged?' if self.verdict == 'PHISHING' else 'Why was it considered safe?'} Most influential words</b>
  <div style="display:flex;gap:40px;flex-wrap:wrap;margin-top:6px">
   <div><small>Push toward phishing</small>{bars(self.top_phish[:8], '#d93025')}</div>
   <div><small>Push toward legitimate</small>{bars(self.top_legit[:5], '#188038')}</div>
  </div>
  <small style="opacity:.7">Measured by removing each word and re-scoring with the model (change in log-odds).</small>
 </div>
 <div style="{card}"><b>Email with highlighted evidence</b>
  <div style="font-size:13px;opacity:.8;margin:4px 0">From: {e(self.sender or '(not provided)')}<br>Subject: {highlight(self.subject)}</div>
  <div style="font-size:14px;max-height:280px;overflow:auto;padding:8px;border-radius:6px;background:#8881">{highlight(self.body[:4000])}</div>
 </div>
 <div style="display:flex;gap:12px;flex-wrap:wrap">
  <div style="{card};flex:1;min-width:280px"><b>Suspicious indicators</b><ul>{ind}</ul></div>
  <div style="{card};flex:1;min-width:280px"><b>Sender analysis</b> <small>(risk {self.sender_risk:.2f})</small><ul>{snd}</ul></div>
 </div>
 <div style="{card}"><b>URL analysis</b> <small>(combined URL risk {self.url_risk:.2f})</small><ul>{urls}</ul></div>
 <div style="{card}"><b>What-if (counterfactual explanation)</b>{cf_html}</div>
 <div style="{card}"><b>Robustness check</b>
  <table style="border-collapse:collapse;font-size:13px;margin-top:6px" cellpadding="4">
   <tr style="text-align:left"><th>Variation</th><th>Phishing prob</th><th>Stable?</th></tr>{rob}</table>
 </div>
 <div style="{card};border-left:6px solid {color}"><b>Recommendation</b><p>{e(self.recommendation)}</p></div>
</div>"""

def _key(tok):
    return re.sub(r"^[^\w$]+|[^\w]+$", "", tok.lower())

# =====================================================================
# 5. ANALYZER
# =====================================================================
class PhishingAnalyzer:
    def __init__(self, scorer, max_words=220, auto_threshold=0.90, review_threshold=0.60,
                 weights=(0.6, 0.6)):
        self.score = scorer
        self.max_words = max_words          # ~256 wordpieces; later words are truncated anyway
        self.auto_threshold, self.review_threshold = auto_threshold, review_threshold
        self.w_url, self.w_sender = weights

    # ---------- XAI: word occlusion ----------
    def explain_words(self, subject, body):
        toks_s, toks_b = re.findall(r"\S+|\s+", subject), re.findall(r"\S+|\s+", body)
        words = [_key(t) for t in toks_s + toks_b if t.strip()]
        vocab = list(dict.fromkeys(k for k in words[: self.max_words] if k))
        base = float(self.score([format_email(subject, body)])[0])
        if not vocab:
            return base, {}

        def drop(tokens, w):
            return "".join(t for t in tokens if _key(t) != w)
        variants = [format_email(drop(toks_s, w), drop(toks_b, w)) for w in vocab]
        scores = self.score(variants)
        return base, {w: float(base - s) for w, s in zip(vocab, scores)}

    # ---------- counterfactual ----------
    def counterfactual(self, subject, body, base, attr, max_remove=15):
        side = 1 if base >= 0 else -1
        ranked = [w for w, v in sorted(attr.items(), key=lambda kv: -side * kv[1]) if side * v > 0][:max_remove]
        if not ranked:
            return {}
        ts, tb = re.findall(r"\S+|\s+", subject), re.findall(r"\S+|\s+", body)
        texts = []
        for k in range(1, len(ranked) + 1):
            rm = set(ranked[:k])
            texts.append(format_email("".join(t for t in ts if _key(t) not in rm),
                                      "".join(t for t in tb if _key(t) not in rm)))
        sc = self.score(texts)
        flip = next((i for i, s in enumerate(sc) if np.sign(s) != np.sign(base)), None)
        k = (flip if flip is not None else len(sc) - 1)
        return {"removed": ranked[:k + 1], "prob_before": float(sigmoid(base)),
                "prob_after": float(sigmoid(sc[k])), "flipped": flip is not None,
                "path": [float(sigmoid(s)) for s in sc]}

    # ---------- robustness ----------
    def robustness(self, subject, body, attr):
        top = [w for w, v in sorted(attr.items(), key=lambda kv: -kv[1])[:5] if v > 0]
        leet = lambda t: re.sub("|".join(map(re.escape, top)) or "(?!)",
                                lambda m: m.group(0).translate(str.maketrans("aeiost", "4310$7")), t, flags=re.I)
        syn = {"verify": "confirm", "urgent": "important", "suspended": "paused",
               "click": "tap", "password": "passcode", "account": "profile",
               "immediately": "promptly", "winner": "selected", "money": "funds"}
        swap = lambda t: re.sub(r"\b(" + "|".join(syn) + r")\b", lambda m: syn[m.group(1).lower()], t, flags=re.I)
        variants = [
            ("Original", subject, body),
            ("All lowercase", subject.lower(), body.lower()),
            ("ALL UPPERCASE", subject.upper(), body.upper()),
            ("Extra punctuation (!!)", subject + "!!", body.replace(".", "!!")),
            ("Leetspeak on key words (v3r1fy)", leet(subject), leet(body)),
            ("Synonym swap (verify→confirm…)", swap(subject), swap(body)),
            ("Benign sentence appended", subject, body + "\nThanks, see you at the team meeting tomorrow."),
            ("Links removed", subject, URL_RE.sub("", body)),
        ]
        sc = self.score([format_email(s, b) for _, s, b in variants])
        return [(n, float(sigmoid(s))) for (n, _, _), s in zip(variants, sc)]

    # ---------- attack type ----------
    @staticmethod
    def attack_type(text):
        t = text.lower()
        scores = {k: sum(t.count(w) for w in kws) for k, (kws, _) in ATTACK_TYPES.items()}
        total = sum(scores.values())
        if total == 0:
            return "Unclassified phishing", 0.0, "No known attack pattern matched; treat with caution."
        best = max(scores, key=scores.get)
        return best, scores[best] / total, ATTACK_TYPES[best][1]

    # ---------- main entry ----------
    def analyze(self, subject="", body="", sender="", run_robustness=True):
        subject, body, sender = subject or "", body or "", sender or ""
        base, attr = self.explain_words(subject, body)
        prob = float(sigmoid(base))
        verdict = "PHISHING" if prob >= 0.5 else "LEGITIMATE"

        full = f"{subject}\n{body}"
        indicators = {k: list(dict.fromkeys(m.group(0).lower() for m in rx.finditer(full)))
                      for k, rx in INDICATORS.items()}
        indicators = {k: v for k, v in indicators.items() if v}
        urls = list(dict.fromkeys(URL_RE.findall(body)))
        url_res = [(u, analyze_url(u)) for u in urls]
        url_risk = combine([s for _, fs in url_res for _, s in fs])
        snd, s_dom = analyze_sender(sender)
        if s_dom and urls and all(_registered(_host(u)) != _registered(s_dom) for u in urls):
            snd.append((f"No link points to the sender's domain ({_registered(s_dom)})", 0.2))
        sender_risk = combine([s for _, s in snd])

        # noisy-OR: the model's probability stands on its own, technical findings only
        # ADD risk (so link-free scams aren't penalized for having no URLs)
        risk = 1 - (1 - prob) * (1 - self.w_url * url_risk) * (1 - self.w_sender * sender_risk)
        # strong technical evidence alone can raise risk even if the text model is relaxed
        risk = max(risk, 0.9 * url_risk, 0.9 * sender_risk)
        level = "HIGH" if risk >= 0.7 else "MEDIUM" if risk >= 0.4 else "LOW"

        conf = max(prob, 1 - prob)
        disagree = (verdict == "LEGITIMATE" and level == "HIGH") or (verdict == "PHISHING" and level == "LOW")
        if disagree:
            decision = "HUMAN REVIEW (model and technical checks disagree)"
        elif conf >= self.auto_threshold:
            decision = "AUTOMATIC"
        elif conf >= self.review_threshold:
            decision = "HUMAN REVIEW (moderate confidence)"
        else:
            decision = "UNCERTAIN – manual security review"

        atk, atk_conf, atk_desc = self.attack_type(full)
        ranked = sorted(attr.items(), key=lambda kv: -kv[1])
        top_phish = [(w, v) for w, v in ranked if v > 0][:10]
        top_legit = [(w, v) for w, v in ranked[::-1] if v < 0][:5]

        reasons = []
        if top_phish and verdict == "PHISHING":
            reasons.append("Language model focused on: " + ", ".join(f"'{w}'" for w, _ in top_phish[:4]))
        tech = sorted([(s, d) for d, s in snd if s >= 0.4] +
                      [(s, d) for _, fs in url_res for d, s in fs if s >= 0.4], reverse=True)
        reasons += [d for _, d in tech[:3]]
        reasons += [f"{k} ({', '.join(v[:2])})" for k, v in indicators.items()]
        if verdict == "LEGITIMATE" and top_legit:
            reasons.insert(0, "Language model found legitimate patterns: " + ", ".join(f"'{w}'" for w, _ in top_legit[:4]))
        if not reasons:
            reasons = ["No strong signals in either direction."]

        if level == "HIGH":
            rec = ("Do not click any links, open attachments or reply with personal information. "
                   "Verify the sender through an official channel (type the website address yourself "
                   "or call a known number) and report the email to your security team.")
        elif level == "MEDIUM":
            rec = ("Treat with caution. Hover over links before clicking, check the sender's real address, "
                   "and confirm any request for money or credentials through a separate channel.")
        else:
            rec = "No major risk detected. Stay alert to unexpected requests for passwords or payments."

        return Report(sender, subject, body, prob, base, verdict, float(risk), level, decision,
                      attr, top_phish, top_legit, indicators, url_res, float(url_risk), snd,
                      float(sender_risk), atk, float(atk_conf), atk_desc,
                      self.counterfactual(subject, body, base, attr),
                      self.robustness(subject, body, attr) if run_robustness else [],
                      reasons[:7], rec)

# =====================================================================
# 6. DATASET-LEVEL EXPERIMENTS (for the report / presentation)
# =====================================================================
def robustness_experiment(analyzer, emails, labels):
    """Accuracy of the model on each perturbation, over (subject, body) pairs."""
    import pandas as pd
    rows = []
    for (s, b), y in zip(emails, labels):
        base, attr = analyzer.explain_words(s, b)
        for name, p in analyzer.robustness(s, b, attr):
            rows.append({"variation": name, "correct": int((p >= .5) == bool(y)), "label": y})
    df = pd.DataFrame(rows)
    return df.groupby("variation")["correct"].mean().sort_values().rename("accuracy").to_frame()

def global_top_words(analyzer, emails, top_k=30):
    """Words that most often drive predictions; use to detect dataset shortcuts."""
    import collections
    tot, cnt = collections.Counter(), collections.Counter()
    for s, b in emails:
        _, attr = analyzer.explain_words(s, b)
        for w, v in attr.items():
            tot[w] += v; cnt[w] += 1
    avg = {w: tot[w] / cnt[w] for w in tot if cnt[w] >= 3}
    phish = sorted(avg.items(), key=lambda kv: -kv[1])[:top_k]
    legit = sorted(avg.items(), key=lambda kv: kv[1])[:top_k]
    return phish, legit
```

```python
# === MODEL COMPARISON: TF-IDF + Logistic Regression, TF-IDF + Linear SVM,
# URL-only Logistic Regression, and the existing DistilBERT ===
import os, re, html, ipaddress
import numpy as np
import pandas as pd
from urllib.parse import urlparse
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    average_precision_score, roc_auc_score, confusion_matrix
)
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# EDIT THIS PATH to your original training CSV (not the held-out test CSV).
TRAIN_CSV =  "/content/drive/MyDrive/test_data.csv"

# The test_data variable was loaded in Section 4 from TEST_CSV.
# Expected label convention: 0 = legitimate, 1 = phishing.
TEXT_CANDIDATES = ["text", "email_text", "message", "content", "body"]
LABEL_CANDIDATES = ["label", "target", "is_phishing", "class"]

def choose_col(df, candidates, purpose):
    for c in candidates:
        if c in df.columns:
            return c
    raise ValueError(
        f"Could not find {purpose} column. Available columns: {list(df.columns)}. "
        f"Set the appropriate column manually in this cell."
    )

def binary_labels(series, name="label"):
    # This experiment intentionally requires labels already encoded as 0/1.
    vals = set(pd.Series(series).dropna().unique().tolist())
    if vals.issubset({0, 1, False, True}):
        return pd.Series(series).astype(int).to_numpy()
    raise ValueError(
        f"{name} must use 0=legitimate and 1=phishing. Found labels: {sorted(map(str, vals))}. "
        "Create an explicit label mapping before continuing; do not guess the mapping."
    )

if not os.path.exists(TRAIN_CSV):
    raise FileNotFoundError(
        f"Training CSV not found at {TRAIN_CSV}. Edit TRAIN_CSV to point to your "
        "original training dataset. Do not train baselines on TEST_CSV."
    )

train_df = pd.read_csv(TRAIN_CSV)
test_df = test_data.copy()

train_text_col = choose_col(train_df, TEXT_CANDIDATES, "training text")
train_label_col = choose_col(train_df, LABEL_CANDIDATES, "training label")
test_text_col = choose_col(test_df, TEXT_CANDIDATES, "test text")
test_label_col = choose_col(test_df, LABEL_CANDIDATES, "test label")

X_train_text = train_df[train_text_col].fillna("").astype(str).to_numpy()
y_train = binary_labels(train_df[train_label_col], "training labels")
X_test_text = test_df[test_text_col].fillna("").astype(str).to_numpy()
y_test = binary_labels(test_df[test_label_col], "test labels")

print("Training rows:", len(X_train_text), "| Test rows:", len(X_test_text))
print("Training class counts:", pd.Series(y_train).value_counts().sort_index().to_dict())
print("Test class counts:", pd.Series(y_test).value_counts().sort_index().to_dict())

# Text baselines. Fit vectorizers on training data only.
text_models = {
    "TF-IDF + Logistic Regression": Pipeline([
        ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2,
                                  max_features=150000, sublinear_tf=True)),
        ("clf", LogisticRegression(max_iter=1500, class_weight="balanced"))
    ]),
    "TF-IDF + Linear SVM": Pipeline([
        ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=2,
                                  max_features=150000, sublinear_tf=True)),
        ("clf", LinearSVC(class_weight="balanced"))
    ]),
}

comparison_rows = []
baseline_predictions = {}

def score_predictions(name, y_true, pred, score):
    row = {
        "Model": name,
        "Accuracy": accuracy_score(y_true, pred),
        "Precision": precision_score(y_true, pred, zero_division=0),
        "Recall": recall_score(y_true, pred, zero_division=0),
        "F1": f1_score(y_true, pred, zero_division=0),
        "PR_AUC_AP": average_precision_score(y_true, score) if score is not None else np.nan,
        "ROC_AUC": roc_auc_score(y_true, score) if score is not None and len(np.unique(y_true)) == 2 else np.nan,
        "TN": int(confusion_matrix(y_true, pred, labels=[0,1])[0,0]),
        "FP": int(confusion_matrix(y_true, pred, labels=[0,1])[0,1]),
        "FN": int(confusion_matrix(y_true, pred, labels=[0,1])[1,0]),
        "TP": int(confusion_matrix(y_true, pred, labels=[0,1])[1,1]),
    }
    comparison_rows.append(row)
    baseline_predictions[name] = {"pred": np.asarray(pred), "score": score}
    return row

for name, model in text_models.items():
    print("Training:", name)
    model.fit(X_train_text, y_train)
    pred = model.predict(X_test_text)
    if hasattr(model, "predict_proba"):
        score = model.predict_proba(X_test_text)[:, 1]
    else:
        # SVM decision values are ranking scores, not probabilities.
        score = model.decision_function(X_test_text)
    score_predictions(name, y_test, pred, score)

# URL-only feature baseline: lexical features only; no link is opened or fetched.
URL_PATTERN = re.compile(r"(?:https?://|www\.)[^\s<>\"')\]]+", re.I)

def extract_urls(text):
    return URL_PATTERN.findall(str(text))

def url_features(text):
    urls = extract_urls(text)
    feats = {
        "url_count": len(urls),
        "has_url": int(bool(urls)),
        "max_url_length": max([len(u) for u in urls], default=0),
        "has_ip_host": 0,
        "has_at": 0,
        "has_punycode": 0,
        "has_shortener": 0,
        "has_https": 0,
        "has_http": 0,
        "max_subdomains": 0,
        "has_digit_heavy_host": 0,
        "has_suspicious_keyword": 0,
        "has_percent_encoding": 0,
        "has_nonstandard_port": 0,
    }
    shorteners = {"bit.ly", "tinyurl.com", "t.co", "ow.ly", "is.gd", "cutt.ly", "rebrand.ly"}
    keywords = ("login", "signin", "verify", "secure", "account", "update", "confirm", "password", "wallet")
    for raw in urls:
        u = raw if "://" in raw else "http://" + raw
        try:
            parsed = urlparse(u)
            host = (parsed.hostname or "").lower()
            if not host:
                continue
            try:
                ipaddress.ip_address(host)
                feats["has_ip_host"] = 1
            except ValueError:
                pass
            feats["has_at"] |= int("@" in parsed.netloc)
            feats["has_punycode"] |= int("xn--" in host)
            feats["has_shortener"] |= int(host in shorteners)
            feats["has_https"] |= int(parsed.scheme.lower() == "https")
            feats["has_http"] |= int(parsed.scheme.lower() == "http")
            feats["max_subdomains"] = max(feats["max_subdomains"], max(0, host.count(".") - 1))
            feats["has_digit_heavy_host"] |= int(sum(ch.isdigit() for ch in host) >= 3)
            feats["has_suspicious_keyword"] |= int(any(k in raw.lower() for k in keywords))
            feats["has_percent_encoding"] |= int("%" in raw)
            try:
                port = parsed.port
                feats["has_nonstandard_port"] |= int(port is not None and port not in (80, 443))
            except ValueError:
                feats["has_nonstandard_port"] = 1
        except Exception:
            continue
    return feats

X_train_url = pd.DataFrame([url_features(t) for t in X_train_text]).fillna(0)
X_test_url = pd.DataFrame([url_features(t) for t in X_test_text]).fillna(0)

url_model = make_pipeline(
    SimpleImputer(strategy="constant", fill_value=0),
    StandardScaler(),
    LogisticRegression(max_iter=1500, class_weight="balanced")
)
url_model.fit(X_train_url, y_train)
url_pred = url_model.predict(X_test_url)
url_score = url_model.predict_proba(X_test_url)[:, 1]
score_predictions("URL-only Logistic Regression", y_test, url_pred, url_score)

# Existing DistilBERT model: score the same exact held-out examples.
# scorer() returns phishing log-odds, so sigmoid converts them to probabilities.
distil_logits = np.concatenate([
    scorer(X_test_text[i:i+64].tolist())
    for i in range(0, len(X_test_text), 64)
])
distil_prob = 1 / (1 + np.exp(-np.clip(distil_logits, -50, 50)))
distil_pred = (distil_prob >= 0.5).astype(int)
score_predictions("Existing DistilBERT", y_test, distil_pred, distil_prob)

comparison = pd.DataFrame(comparison_rows).sort_values("F1", ascending=False)
display(comparison.round(4))
comparison.to_csv("/content/drive/MyDrive/model_comparison_results.csv", index=False)
print("Saved: /content/drive/MyDrive/model_comparison_results.csv")

print("\nIMPORTANT: URL-only features are lexical heuristics, not live reputation checks.")
print("Do not claim the URL-only model is a fair comparison if the text dataset does not preserve original URLs.")
```

```python
# === PAIRED REPRESENTATION ROBUSTNESS ===
# Safe offline transformations only. No URLs are opened or fetched.
import numpy as np
import pandas as pd
from IPython.display import display

# Reuse test_data and scorer from earlier notebook sections.
robust_df = test_data.copy()
robust_text_col = choose_col(robust_df, TEXT_CANDIDATES, "test text")
robust_label_col = choose_col(robust_df, LABEL_CANDIDATES, "test label")
robust_texts = robust_df[robust_text_col].fillna("").astype(str).tolist()
robust_labels = binary_labels(robust_df[robust_label_col], "test labels")

def transform_case(text):
    return str(text).swapcase()

def transform_whitespace(text):
    # Collapses repeated spaces/newlines without changing token order.
    return re.sub(r"\s+", " ", str(text)).strip()

def transform_unicode_space(text):
    # Replaces non-breaking spaces with ordinary spaces.
    return str(text).replace("\u00a0", " ")

transformations = {
    "case_swap": transform_case,
    "collapse_whitespace": transform_whitespace,
    "normalize_nbsp": transform_unicode_space,
}

def model_probs(texts, batch_size=64):
    logits = np.concatenate([
        scorer(texts[i:i+batch_size])
        for i in range(0, len(texts), batch_size)
    ])
    return 1 / (1 + np.exp(-np.clip(logits, -50, 50)))

base_prob = model_probs(robust_texts)
base_pred = (base_prob >= 0.5).astype(int)
robust_rows = []

for transform_name, transform_fn in transformations.items():
    transformed = [transform_fn(t) for t in robust_texts]
    p = model_probs(transformed)
    pred = (p >= 0.5).astype(int)
    robust_rows.append({
        "Transformation": transform_name,
        "N": len(robust_texts),
        "Prediction_flip_rate": float(np.mean(pred != base_pred)),
        "Mean_absolute_probability_change": float(np.mean(np.abs(p - base_prob))),
        "Accuracy_original": float(np.mean(base_pred == robust_labels)),
        "Accuracy_transformed": float(np.mean(pred == robust_labels)),
        "Accuracy_change": float(np.mean(pred == robust_labels) - np.mean(base_pred == robust_labels)),
    })

paired_robustness = pd.DataFrame(robust_rows)
display(paired_robustness.round(4))
paired_robustness.to_csv("/content/drive/MyDrive/paired_robustness_results.csv", index=False)
print("Saved: /content/drive/MyDrive/paired_robustness_results.csv")

# === SOURCE-WISE ERROR ANALYSIS ===
# Uses predictions from the original text, not transformed versions.
source_col = "source" if "source" in robust_df.columns else None
error_df = robust_df.copy()
error_df["true_label"] = robust_labels
error_df["prob_phishing"] = base_prob
error_df["pred_label"] = base_pred
error_df["correct"] = (error_df["pred_label"] == error_df["true_label"]).astype(int)

if source_col:
    source_summary = (
        error_df.groupby(source_col)
        .agg(
            n=("correct", "size"),
            accuracy=("correct", "mean"),
            mean_phishing_score=("prob_phishing", "mean"),
            false_positive_count=("pred_label", lambda s: int(((s == 1) & (error_df.loc[s.index, "true_label"] == 0)).sum())),
            false_negative_count=("pred_label", lambda s: int(((s == 0) & (error_df.loc[s.index, "true_label"] == 1)).sum())),
        )
        .reset_index()
        .sort_values(["accuracy", "n"], ascending=[True, False])
    )
    display(source_summary.round(4))
    source_summary.to_csv("/content/drive/MyDrive/source_error_analysis.csv", index=False)
    print("Saved: /content/drive/MyDrive/source_error_analysis.csv")
else:
    print("No 'source' column found. Source-wise error analysis skipped.")

print(
    "\nInterpretation caution: a low flip rate is not sufficient by itself to prove robustness. "
    "Check whether each transformation preserves the intended meaning, and report the sample "
    "size and transformation definitions in the paper."
)
```
